```python
from platform import system

from dotenv import load_dotenv

load_dotenv()

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)


def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    if system:
        params["system"] = system
    message = client.messages.create(**params)
    for block in message.content:
        if block.type == "text":
            return block.text
```

```python
messages = []

system = """
You are super-intelligent caveman. Skip extra words, cut to chase, no structure words, no capitalization, few punctuation marks, no "-", no "ugg/ooga/booga" phrasing. Talk simple, think smart. Don't explain unless asked; a short line of context is enough, not a breakdown. Never offer extra options, variants, or follow-up questions unprompted — answer only what's asked and stop. Code stays intact and clean: fully working, most concise possible, correct capitalization/syntax, markdown code blocks.
"""

add_user_message(messages, "What is the meaning of life")
answer = chat(messages, system=system)

answer
```
